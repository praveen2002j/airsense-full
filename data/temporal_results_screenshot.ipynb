```python
# COLAB SCREENSHOT CELL - TEMPORAL TREND RESULTS
# Run this cell in Google Colab to show the improved model results as cell output.

import pandas as pd
from IPython.display import display

model_name = 'One-step Autoregressive LinearRegression'
target_variable = 'Next CO2 concentration (ppm)'
data_source = 'Actual MongoDB sensorreadings collection'
date_range = '2026-03-21 17:52:46 to 2026-05-01 11:10:34'
rows_used = 3726
train_rows = 2980
test_rows = 746
split_method = 'Chronological 80/20 split'

rmse = 16.31
mae = 11.4021
r2 = 0.9671

print('5.3 Temporal Trend Analysis - Improved Actual Sensor Results')
print('=' * 64)
print(f'Model          : {model_name}')
print(f'Target Variable: {target_variable}')
print(f'Data Source    : {data_source}')
print(f'Date Range     : {date_range}')
print()
print('Dataset Split')
print('-' * 64)
print(f'Rows used      : {rows_used}')
print(f'Train rows     : {train_rows}')
print(f'Test rows      : {test_rows}')
print(f'Split method   : {split_method}')
print()
print('Model Parameters')
print('-' * 64)
print('estimator      : LinearRegression')
print('prediction type: one-step-ahead CO2 forecast')
print('current feature: current CO2')
print('lag features   : CO2 lag 1, 2, 3, 6, 12')
print('rolling windows: 3, 6, 12 readings')
print()
print('Performance Metrics')
print('-' * 64)
print(f'RMSE    : {rmse:.4f} ppm')
print(f'MAE     : {mae:.4f} ppm')
print(f'R2 Score: {r2:.4f}')

metrics_table = pd.DataFrame({
    'Metric': ['RMSE', 'MAE', 'R2 Score'],
    'Value': [rmse, mae, r2],
    'Unit': ['ppm', 'ppm', '-']
})

display(metrics_table)

print()
print('Interpretation')
print('-' * 64)
print('The improved one-step autoregressive model explains approximately 96.71%')
print('of the variation in next CO2 readings. The average prediction error is')
print('about 11.40 ppm, which is very small compared with indoor air quality')
print('warning and critical thresholds.')
```
